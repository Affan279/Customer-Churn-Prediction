```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder


categorical_cols = [

     'type_policy',
     'type_policy_dg',
     'type_product',
     'reimbursement',
     'new_business',
     'distribution_channel',
     'gender'
    ]

numerical_cols = [
     'period',
     'year_effect_insured',
     'year_effect_policy',
     'exposure_time',
     'lapse',
     'seniority_insured',
     'seniority_policy',
     'age',
     'premium',
     'cost_claims_year',
     'n_medical_services',
     'n_insured_pc',
     'n_insured_mun',
     'n_insured_prov',
]

preprocessor = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_cols),
        (
            "num",
            "passthrough",
            numerical_cols
        )
    ]
)
```

```python
X_encoded = preprocessor.fit_transform(insu_df[categorical_cols+ numerical_cols])
```

```python
categorical_cols = [

     'type_policy',
     'type_policy_dg',
     'type_product',
     'reimbursement',
     'new_business',
     'distribution_channel',
     'gender'
    ]

numerical_cols = [
     'period',
     'year_effect_insured',
     'year_effect_policy',
     'exposure_time',
     'lapse',
     'seniority_insured',
     'seniority_policy',
     'age',
     'premium',
     'cost_claims_year',
     'n_medical_services',
     'n_insured_pc',
     'n_insured_mun',
     'n_insured_prov',
]

missing = [
 'IICIMUN',
 'IICIPROV',
 'C_GI',
 'C_II',
 'C_IE_P',
 'C_IE_S',
 'C_IE_T',
 'C_GE_P',
 'C_GE_S',
 'C_GE_T'
 ]
```

```python
insu_df = insu_df[categorical_cols+ numerical_cols+missing]

filled_insu_df = insu_df[insu_df.notna().all(axis=1)]
```

```python
filled_insu_df.dtypes.loc[missing]
```

```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
from sklearn.ensemble import ExtraTreesRegressor

imputer = IterativeImputer(
    estimator=ExtraTreesRegressor(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    ),
    max_iter=10,
    random_state=42,
    initial_strategy="median"
)
```

```python
result = imputer.fit(filled_insu_df[missing])
```
